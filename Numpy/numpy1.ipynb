```python
# phase 2
# indexing and slicing
arr=np.array([1,2,3,4,5])
print("first element: ",arr[0])
print("last element: ",arr[-1])
print("slicing", arr[2:7])
print("second",arr[1:8:2])
print("reverse",arr[::-1])
print("negative",arr[-3])

# 2d array
print("2d array")
arr=np.array([[1,2,3],[4,5,6]])

print("specfic element",arr[1,2])
print("Entire row",arr[1])
print("entire col",arr[:,1])
# sorting

unsorted=np.array([1,7,8,9,5,7,2,9,2])
print("sorted array",np.sort(unsorted) )

arr2=np.array([[3,1],[1,2],[2,3]])
print("sorted 2d array by col",np.sort(arr2,axis=0))
print("sorted 2d array by row",np.sort(arr2,axis=1))

#filter

number=np.array([1,2,3,4,5,6,7,9,0])
even_number=number[number%2==0]
print("even numbers",even_number)

# filter with mask
mask=number>5
print("mask",number[mask])

# fancy indexing and np.where() clause


indicies=[0,2,4]
print(number[indicies])


where_result=np.where(number%2==0)
print("where clause",where_result)

#condition array
condition_array=np.where(number%2==0,"even","odd")
print("condition array",condition_array)


# adding and removing of the data

arr1=np.array([1,2,3])
arr2=np.array([4,5,6])

concatenated=np.concatenate((arr1,arr2))
print("concatenated array",concatenated)


# array comatibality
a=np.array([1,2,3])
b=np.array([4,5,6])
c=np.array([7,8,9])

print("compatablity",a.shape==b.shape)
original=np.array([[1,2],[3,4]])
new_row=np.array([[5,6]])

print("new vstack")

with_new_row=np.vstack((original,new_row))
print(with_new_row)

new_col=np.array([[7],[8]])

print("new hstack")
new_col=np.array([[7],[8]])
with_new_row=np.hstack((original,new_col))
print(with_new_row)

arr=np.array([1,2,3,4,5])
deleted=np.delete(arr,2)
print(deleted)



```

```python
# phase 3
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
import matplotlib.pyplot as plt

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("sales data")
print(sales_data)
print(sales_data.shape)
print(sales_data.ndim)
print(sales_data.dtype)
print("sample data",sales_data[:,1:])

total_sales=np.sum(sales_data[:,1:],axis=1)
print("total sales",total_sales)

yearly_total=np.sum(sales_data[:,1:],axis=0)
print("yearly total",yearly_total)

# minimum sale of the resturant
min_sales=np.min(sales_data[:,1:],axis=1)
print("minimum sales",min_sales)

#maximum sales of the customer
max_sales=np.max(sales_data[:,1:],axis=0)
print("maximum sales",max_sales)

# average sales of the customer
avg_sales=np.mean(sales_data[:,1:],axis=1)
print("average sales",avg_sales)

# cumulative sales
cumulative_sales=np.cumsum(sales_data[:,1:],axis=1)
print("cumulative sales",cumulative_sales)

plt.figure(figsize=(10,6))
plt.plot(sales_data[0,1:],label="Paradise Biryani")
plt.title("Sales Analysis")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()
plt.show()






```

```python
# phase 4
# broadcasting


import numpy as np
import matplotlib.pyplot as plt

# Create NumPy arrays
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.random.rand(3, 3)
array3 = np.zeros((4, 4))

# Save arrays to files
np.save('array1.npy', array1)
np.save('array2.npy', array2)
np.save('array3.npy', array3)

# Load and print array1
loaded_array1 = np.load('array1.npy')
print("Loaded Array 1:")
print(loaded_array1)

# Load and display NumPy logo
try:
    logo = np.load('numpy-logo.npy')

    # Display original and dark logo
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(logo)
    plt.title("NumPy Logo")
    plt.grid(False)
    plt.axis('off')

    # Invert colors
    dark_logo = 1 - logo

    plt.subplot(1, 2, 2)
    plt.imshow(dark_logo)
    plt.title("NumPy Dark Logo")
    plt.grid(False)
    plt.axis('off')

    plt.tight_layout()
    plt.show()

except FileNotFoundError:
    print("NumPy logo file not found. Please check the file path.")

```
